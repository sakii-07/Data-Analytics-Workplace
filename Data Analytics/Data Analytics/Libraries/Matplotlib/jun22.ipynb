```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb','Mar','Apr','May']
sales = [10000,15000,12000,18000,16000]
plt.plot(months,sales)
plt.show()
```

```python
# line properties
months = ['Jan', 'Feb','Mar','Apr','May']
sales = [10000,15000,12000,18000,16000]
plt.plot(months,sales, color = 'hotpink',linestyle="-.", lw=2)
plt.show()
```

```python
# marker properties
months = ['Jan', 'Feb','Mar','Apr','May']
sales = [10000,15000,12000,18000,16000]
plt.plot(months,sales, marker = "p",mfc="red",mec="black",ms=10)
plt.show()
```

```python
# grid properties
months = ['Jan', 'Feb','Mar','Apr','May']
sales = [10000,15000,12000,18000,16000]
plt.plot(months,sales)
plt.grid(color="green",ls="--",lw=1)
plt.show()
```

```python
# label properties
months = ['Jan', 'Feb','Mar','Apr','May']
sales = [10000,15000,12000,18000,16000]
plt.plot(months,sales,linestyle="-.")
plt.xlabel("Months",color="red",loc="center")
plt.ylabel("Sales",color="red",loc="center")
plt.title("Sales analysis",color="green",loc="center")
plt.show()
```

```python
students = pd.read_csv("students_data.csv")
students
```

```python
mon_gp = students.groupby(by="month")
sales = mon_gp['fees_paid'].sum()
plt.plot(sales.index,sales.values, linestyle="-.",marker="p",color="purple")
plt.show()
```

```python
# Bar graph
course = ['DA','DS','WD']
addm = [130,150,90]
plt.bar(course,addm,width=0.5,color='gray')
plt.show()
```

```python
# Horizontal Bar graph
course = ['DA','DS','WD']
addm = [130,150,90]
plt.Figure(figsize=(3,4))
plt.barh(course,addm,height=0.5,color='orange')
plt.ylabel("Courses", fontdict={'family':'sans-serif',
                                'color':'darkred','weight':'bold','size':14})
plt.xlabel("Admission", fontdict={'family':'sans-serif',
                                'color':'darkred','weight':'bold','size':14})
plt.show()
```

```python
course_gp = students.groupby(by="course")
data = course_gp['course'].count() # we also use values_count()
plt.bar(data.index,data.values,color="skyblue",width=0.5)
plt.show()
```

```python
# Pie chart
products =['Laptop','Mobile','Tv']
sales = [10,7,3]
plt.pie(sales,labels=products,explode=[0,0.2,0.2], colors=['skyblue','orange','hotpink'])
plt.legend()
plt.show()
```

```python
sales = course_gp['fees_paid'].sum()
plt.pie(sales.values,labels=sales.index,colors=['skyblue','orange','purple'],explode=[0.05,0.01,0.2],
        autopct="%1.2f%%",shadow=True)
plt.legend()
plt.show()
```

```python
# Histogram 
wt = [41,45,48,57,59,64,68,70,78,82,84,90,93,95,97]
plt.hist(wt,bins=4)
plt.show()
```
