```python
def perf(rating):
    if rating>4:
        return "Excellent"
    else:
        return "Good"
```

```python
emp_df['performance'] = emp['rating'].apply(perf)
emp_df
```

```python
def increment(salary):
    if salary > 50000:
        new_salary = salary + salary*10/100
    elif salary > 40000:
        new_salary = salary + salary*15/100
    elif salary > 30000:
        new_salary = salary + salary*20/100
    else:
        new_salary = salary + salary*25/100
    return new_salary
```

```python
emp_df['salary'] = emp_df['salary'].apply(increment)
emp_df
```

```python
emp_df['department'].replace("IT",'Operation',inplace=True)
emp_df
```
