```python
st1 = "select * from students"
st2 = "select * from courses"
st3 = "select * from branch"
students = pd.read_sql(st1,conn)
students
```

```python
branch = pd.read_sql(st3,conn)
branch
```

```python
df1 = pd.merge(students,branch,how="inner",on="br_id")
df2 = pd.merge(df1,courses,how="inner",on="course_id")
student_data = df2[['student_id','student_name','br_name','course_name','duration','fees','paid_fee']]
```
