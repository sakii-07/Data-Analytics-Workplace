```python
# numpy library
import numpy as np
```

```python
a0 = np.array(45)
a1 = np.array([34,67,89])
a2 = np.array([[1,2,3],[5,6,7],[3,7,9]])
a3 = np.array([[[1,2,3],[6,7,8]],[[9,6,4],[2,5,8]]])
```

```python
# Indexing
print(a1[1])
```

```python
print(a2)
print()
print(a2[1])
print()
print(a2[1][2])
```

```python
print(a3)
print()
print(a3[1])
print()
print(a3[1][1])
print()
print(a3[1][1][2])
```

```python
# alternative way
print(a3[1,1,2])
print(a3[1][1][2])
```

```python
# Slicing
a1 = np.array([1,2,3,4,5,6,7,8])
print(a1[2:6:2])
print(a1[2:6])# step value = 1
print(a1[:5]) # start with 0
print(a1[3::1]) # end at -1
```

```python
a3 = np.array([[11,22,33,44,55],[1,2,3,4,5],[12,12,12,12,12]])
```

```python
print(a3)
print()
print(a3[1,1:3])
print()
print(a3[0,-2:0:-1])
```

```python
# Arithmetic operations
a1 = np.array([1,2,3,4,5])
a2 = np.array([8,7,6,5,4])

print(a1+a2)
print(a1*a2)
print(a1/a2)
```

```python
# Aggrigate function
np.sum(a1)
np.min(a1)
np.max(a1)
np.std(a1)
```

```python
# reshape - change dimentions
print(a1)
a = a1.reshape((2,4))
```

```python
# flatten() - convert multi-dimentional array to one dimentional array -- retrun copy                                                                                                                                                                                                   -- deep copy
# ravel() - convert multi-dimentional array to one dimentional array -- return view -- shallow copy
# copy() -- deep copy
# view() -- create only view
a = a2.flatten()
print(a)
a[1] = 999
print(a)
print(a1)
```

```python
a = a2.ravel()
print(a)
a[1] = 999
print(a)
print(a1)
```

```python
# Join two or more array using 
# 1) concatenate()
# stack() - axis=0 (row wise) -- axis=1 (column wise)

a1 = np.array([1,2,3,4])
a2 = np.array([6,7,8,9])

np.stack((a1,a2),axis=0)
```

```python
np.stack((a1,a2),axis=1)
```

```python
a1 = np.array([1,2,3,4])
a2 = np.array([6,7,8,9])

np.hstack((a1,a2))
np.vstack((a1,a2))
```

```python
# Where function -- it returns index number
a = np.array([12,24,45,67,78,90,86])
index = np.where(a>50) # returns index number
a[index] # access element by index number
```

```python
a = np.array([12,24,45,67,78,90,86])
print(a[np.where(a%3==0)])
print(a[np.where(a%5==0)])
print(a[np.where((a%3==0) & (a%5==0))])
```

```python
marks = np.array([12,24,45,67,78,90,86])
np.where(marks>40,"Pass","Fail")
```
