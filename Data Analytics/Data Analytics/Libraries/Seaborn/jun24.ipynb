```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

```python
sales_data ={
    'month':['jan','feb','mar','apr'],
    'sales':[10,34,67,45]
}

sales_df = pd.DataFrame(sales_data)
sns.lineplot(data = sales_df, x='month', y='sales',marker='p')
plt.show()
```

```python
sales_data ={
    'month':['jan','feb','mar','apr','jan','feb','mar','apr'],
    'sales':[10,34,67,45,15,24,57,75],
    'branch':['Karvenagar','Karvenagar','Karvenagar','Karvenagar','Deccan','Deccan','Deccan','Deccan']
}

# line plot
sales_df = pd.DataFrame(sales_data)
sns.lineplot(data = sales_df, x='month', y='sales',marker='p')
plt.show()
```

```python
sales_df = pd.DataFrame(sales_data)
sns.lineplot(data = sales_df, x='month', y='sales',marker='p', hue='branch')
plt.show()
```

```python
sales_df = pd.DataFrame(sales_data)
sns.lineplot(data = sales_df, x='month', y='sales',marker='p',estimator="min")
plt.show()
```

```python
sales_df = pd.DataFrame(sales_data)
sns.lineplot(data = sales_df, x='month', y='sales',marker='p', hue='branch', style='branch')
plt.show()
```

```python
# barchart
sns.barplot(sales_df,x='branch',y='sales', hue='month',estimator='sum',palette="Set1", gap=0.3)
plt.legend()
plt.show()
```

```python
sns.get_dataset_names()
tips = sns.load_dataset('tips')
tips
```

```python
# barplot
sns.barplot(data=tips, x='time',y='total_bill',estimator='sum')
plt.show()
```

```python
# countplot
sns.countplot(data=tips, x='sex', hue='smoker', palette='Set1')
plt.show()
```

```python
# scatter - used to find relation between two columns
# 1) positive relation
# 2) negative relation
# 3) no relation

sns.scatterplot(data=tips, x='total_bill',y='tip',hue='time',style='sex')
plt.show()
```

```python
# histogram - shows data distribution
sns.histplot(data=tips, x='total_bill',bins=[10,20,30,40,50],hue='sex')
plt.show()
```

```python
# boxplot - used to find outliers and we see the statatistics properties
sns.boxplot(data=tips, y='total_bill')
plt.show()
```

```python
sns.boxplot(data=tips, y='total_bill', hue='sex')
plt.show()
```
