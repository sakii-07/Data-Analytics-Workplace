```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Employee_Data_20_Records.xlsx')
df
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df['age'] = scaler.fit_transform(df[['age']])
df
```

```python
df['experience'] = scaler.fit_transform(df[['experience']])
df
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
df['salary'] = scaler.fit_transform(df[['salary']])
df
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
df['gender'] = encoder.fit_transform(df[['gender']])
df
```

```python
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['Average','Good','Excellent']])
df['rating'] = encoder.fit_transform(df[['rating']])
df
```

```python
data = pd.get_dummies(df, columns=['department'])
data
```
