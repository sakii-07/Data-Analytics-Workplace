```python
# Task 8: Employee Salary Analysis

# Salary data:

# [25000, 30000, 45000, 50000, 35000, 60000, 70000]

# Find:

# Average salary
# Highest salary
# Lowest salary
# Salaries above average
# Increase all salaries by 10%

salary = np.array([25000, 30000, 45000, 50000, 35000, 60000, 70000])
mean = np.mean(salary)
print(mean)
print(np.max(salary))
print(np.min(salary))
print(salary[mean>salary])
salary = salary+salary*(10/100)
print(salary)
```

```python
# Task 15: Complete NumPy Challenge

# Create a dataset of 500 students' marks using random numbers.

# Calculate:

# Top 10 marks
# Bottom 10 marks
# Average marks
# Students scoring above 75
# Students scoring below 35
# Pass percentage
# Grade distribution (A, B, C, D, F)

marks = np.random.randint(30,100,500)
marks
```

```python
print(np.sort(marks)[-10:])
print()
print(np.sort(marks)[:11])
print(marks[marks>75])
print(marks[marks<35])

pass_per = (np.sum(marks>=35) / len(marks)) * 100
print(pass_per)
```

```python
# grade distribution

A = np.sum(marks>95)
B = np.sum((marks<95) & (marks>85))
C = np.sum((marks<85) & (marks>75))
D = np.sum((marks<75) & (marks>65))
E = np.sum((marks<65) & (marks>55))
F  = np.sum(marks <55)
```
