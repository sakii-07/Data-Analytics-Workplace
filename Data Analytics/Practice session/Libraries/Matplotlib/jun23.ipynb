```python
import matplotlib.pylab as plt
import pandas as pd
```

```python
# Task 1: Branch-wise Student Count

# Objective: Har branch me kitne students hain.

# Chart: Bar Chart

# Columns Used: branch

# Requirements:

# X-axis â†’ Branch Name
# Y-axis â†’ Number of Students
# Add Title
# Add X-axis and Y-axis Labels
# Show Count on Top of Bars

student_df = pd.read_csv('student_data.csv')
branch_gp = student_df.groupby(by='branch')
std_count = branch_gp['branch'].value_counts()
bars = plt.bar(std_count.index, std_count.values)
plt.xlabel('Branch Name')
plt.ylabel('Student Count')
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        str(height),
        ha='center',
        va='bottom'
    )
plt.show()
```

```python
# Task 2: Course-wise Student Distribution

# Objective: Har course me kitne students enrolled hain.

# Chart: Pie Chart

# Columns Used: course_name

# Requirements:

# Show percentage on each slice
# Add title
# Start angle 90Â°
# Add shadow effect

course_gp = student_df.groupby(by='course_name')
std = course_gp['course_name'].value_counts()
plt.pie(std.values,labels = std.index,shadow=True, colors=['orange','hotpink','skyblue','yellow','purple'])
plt.title('Course-wise Student Distribution')
plt.legend()
plt.show()
```

```python
# Task 3: Branch-wise Average Marks

# Objective: Har branch ke students ke average marks calculate karo.

# Chart: Bar Chart

# Columns Used: branch, marks

# Requirements:

# Calculate average marks branch-wise
# Add different colors to bars
# Display average value on each bar

avg = branch_gp['marks'].mean()
bars = plt.bar(avg.index,avg.values,color=['red','purple','green','skyblue','hotpink'])
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        height,
        f"{height:.2f}",
        ha='center',
        va='bottom'
    )
plt.xlabel("Branch Name",color="red")
plt.ylabel("Average Marks",color="red")
plt.show()
```

```python
# Task 4: Course-wise Average Marks

# Objective: Kis course ke students ka performance best hai.

# Chart: Horizontal Bar Chart

# Columns Used: course_name, marks

# Requirements:

# Sort values in descending order
# Add title and labels
# Show average marks
course_gp = student_df.groupby(by='course_name')
avg = course_gp['marks'].mean().sort_values()
plt.barh(avg.index, avg.values, color = "skyblue")
plt.xlabel("Average Marks",color="green")
plt.ylabel("Branch Name",color="green")
plt.title("Branch wise average marks",color="green")
plt.show()
```

```python
# Task 5: Marks Distribution Analysis

# Objective: Students ke marks kis range me jyada hain.

# Chart: Histogram

# Columns Used: marks

# Requirements:

# Use 10 bins
# Add grid
# Add title
# Identify most frequent marks range

plt.hist(student_df['marks'], bins=10,color="Green")
plt.grid()
plt.title("Marks Distribution Analysis",color="Green")
plt.show()
```

```python
# Task 6: Fees vs Marks Relationship

# Objective: Kya fees payment aur marks ke beech koi relationship hai?

# Chart: Scatter Plot

# Columns Used: paid_fees, marks

# Requirements:

# Add title
# Add axis labels
# Analyze Positive, Negative or No Relation
plt.scatter(student_df['paid_fees'], student_df['marks'],color="Green")
plt.xlabel("Paid fees",color="Green")
plt.ylabel("Marks",color="Green")
plt.title("Fees vs Marks Relationship")
plt.show()
```

```python
# Task 7: Remaining Fee Distribution

# Objective: Students ki pending fee distribution dekhna.

# Chart: Histogram

# Columns Used: remaining_fee

# Requirements:

# Use 15 bins
# Add grid
# Find most common remaining fee range
plt.hist(student_df['remaining_fee'],bins = 15)
plt.grid()
plt.show()
```

```python
# Task 8: Branch-wise Total Revenue

# Objective: Har branch ne kitna revenue generate kiya.

# Chart: Bar Chart

# Columns Used: branch, paid_fees

# Requirements:

# Calculate total paid fees
# Sort descending
# Add labels and title
gp = student_df.groupby(by='branch')
fees = gp['paid_fees'].sum().sort_values()
plt.bar(fees.index,fees.values,color="orange")
plt.xlabel("Branch Name",color="orange")
plt.ylabel("Total Revenue",color="orange")
plt.title("Branch-wise Total Revenue",color="orange")
plt.show()
```

```python
# Task 9: Course-wise Revenue Contribution

# Objective: Total revenue me har course ka contribution dekhna.

# Chart: Pie Chart

# Columns Used: course_name, paid_fees

# Requirements:

# Calculate total paid fees course-wise
# Show percentage contribution
# Add title

fees = gp['paid_fees'].sum()
plt.pie(fees.values, labels=fees.index,autopct="%1.2f%%", colors=['orange','hotpink','skyblue','purple','yellow'],
        explode=[0.1,0,0,0,0])
plt.show()
```

```python
# Task 10: Top 10 Students by Marks

# Objective: Highest marks score karne wale students identify karo.

# Chart: Bar Chart

# Columns Used: name, marks

# Requirements:

# Select Top 10 students
# Sort descending
# Rotate student names by 45Â°

top10 = student_df.sort_values(by="marks",ascending=False).head(10)
plt.bar(top10['name'], top10['marks'],color="purple")
plt.xticks(rotation=45)
plt.show()
```

```python
# Task 11: Top 5 Defaulters

# Objective: Jinki remaining fee sabse jyada hai.

# Chart: Bar Chart

# Columns Used: name, remaining_fee

# Requirements:

# Find Top 5 students with highest remaining fee
# Sort descending
# Show remaining fee amount on bars
# Add title and labels

std = student_df.sort_values(by="remaining_fee",ascending=False).head()
plt.bar(std['name'], std['remaining_fee'],color="brown")
plt.xlabel("Student name",color="brown")
plt.ylabel("Remaining fees",color="brown")
plt.title("Top 5 Defaulters",color="brown")
plt.xticks(color="black")
plt.show()
```

```python
# Task 12: Marks Outlier Detection

# Objective: Marks me outliers identify karna.

# Chart: Box Plot

# Columns Used: marks

# Requirements:

# Show median
# Identify outliers
# Add title

plt.boxplot(student_df['marks'], showmeans=True)
plt.title("Distribution of Student Marks with Outliers",color="green")
plt.ylabel("Marks",color="green")
plt.show()
```

```python
# Task 13: Branch-wise Marks Comparison

# Objective: Different branches ke marks compare karna.

# Chart: Multiple Box Plots

# Columns Used: branch, marks

# Requirements:

# One boxplot per branch
# Compare median and spread
# Add title

student_df.boxplot(column='marks', by='branch')
plt.xlabel("Branch",color='green')
plt.ylabel("Marks",color='green')
plt.title("Branch-wise Marks Comparison",color='green')
plt.show()
```

```python
# Task 14: Fee Collection Status

# Objective: Total Fees aur Total Paid Fees compare karna.

# Chart: Bar Chart

# Columns Used: fees, paid_fees

# Requirements:

# Calculate total fees
# Calculate total paid fees
# Compare using two bars

total_fees = student_df['fees'].sum()
total_paid_fees = student_df['paid_fees'].sum()
plt.bar(['total_fees','total_paid_fees'],[total_fees,total_paid_fees])
plt.show()
```

```python
# Task 15: Top 5 Courses by Revenue

# Objective: Sabse jyada revenue generate karne wale courses identify karo.

# Chart: Bar Chart

# Columns Used: course_name, paid_fees

# Requirements:

# Calculate revenue course-wise
# Sort descending
# Display revenue values on bars
gp = student_df.groupby(by='course_name')
fees = gp['paid_fees'].sum().sort_values(ascending=False)
bars = plt.bar(fees.index,fees.values,color="darkgreen")
plt.title("Course-wise Revenue",color="darkgreen")
plt.xlabel("Course",color="darkgreen")
plt.ylabel("Revenue (Paid Fees)",color="darkgreen")
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{int(bar.get_height())}",
        ha='center',
        va='bottom'
    )
```

```python
# Task 16: Student Performance Dashboard

# Objective: Ek figure me multiple insights dikhana.

# Charts Required:

# Branch-wise Student Count
# Course-wise Student Count
# Marks Histogram
# Revenue by Branch

# Requirements:

# Use subplot()
# Figure Size (12,8)
# Proper titles for each chart
course_gp = student_df.groupby(by='course_name')
branch_gp = student_df.groupby(by='branch')
bt_count = branch_gp['branch'].value_counts()
ct_count = course_gp['course_name'].value_counts()
b_fees = branch_gp['paid_fees'].sum()

plt.figure(figsize=(12,8))

plt.subplot(2,2,1)
plt.bar(bt_count.index,bt_count.values)
plt.title("Branch-wise Student Count")

plt.subplot(2,2,2)
plt.bar(ct_count.index, ct_count.values)
plt.title("Course-wise Student Count")

plt.subplot(2,2,3)
plt.hist(student_df['marks'])
plt.title("Marks Histogram")

plt.subplot(2,2,4)
plt.bar(b_fees.index, b_fees.values)
plt.title("Revenue by Branch")
plt.show()
```

```python
# Task 17: Fees vs Remaining Fee Analysis

# Objective: Total fee aur pending fee ka relationship samajhna.

# Chart: Scatter Plot

# Columns Used: fees, remaining_fee

# Requirements:

# Add grid
# Add labels
# Analyze trend

plt.scatter(student_df['fees'],student_df['remaining_fee'])
plt.title("Fees vs Remaining Fee Analysis")
plt.xlabel("Total Fees")
plt.ylabel("Remaining Fee")
plt.grid()
plt.show()
```

```python
# Task 18: Branch-wise Remaining Fee

# Objective: Kis branch me sabse jyada pending fees hai.

# Chart: Bar Chart

# Columns Used: branch, remaining_fee

# Requirements:

# Calculate total remaining fee branch-wise
# Sort descending
# Add value labels

fees = branch_gp['remaining_fee'].sum().sort_values()
plt.bar(fees.index, fees.values,color='orange')
plt.xlabel("Branch name",color='orange')
plt.ylabel("Total remaining fee",color='orange')
plt.show()
```

```python
# Task 19: Course Popularity Analysis

# Objective: Sabse popular course identify karna.

# Chart: Bar Chart

# Columns Used: course_name

# Requirements:

# Count students per course
# Sort descending
# Display count labels

count = course_gp['course_name'].value_counts().sort_values(ascending=False)
bars = plt.bar(count.index, count.values, color='skyblue')
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{int(bar.get_height())}",
        ha='center',
        va='bottom'
    )
plt.show()
```

```python
# Task 20: Student Performance Trend

# Objective: Marks ko ascending order me visualize karna.

# Chart: Line Chart

# Columns Used: marks

# Requirements:

# Sort marks in ascending order
# Add markers
# Add grid
# Add title and labels
std_marks = student_df['marks'].sort_values()
plt.plot(std_marks.values, marker='o',mec='green')
plt.grid()
plt.title("Student Marks (Ascending Order)")
plt.xlabel("Student Index")
plt.ylabel("Marks")
plt.show()
```
