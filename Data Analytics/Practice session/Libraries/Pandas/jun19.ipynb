```python
# Pandas + SQL Practice Task
# ---------------------------
# You have three CSV files:
#     products.csv
#     categories.csv
#     sales.csv
# Task 1: Import CSV Files into MySQL
# Create a database named product_db.
# Create the following tables:
#     products
#     categories
#     sales
# Import all three CSV files into their respective tables.
# Verify that the data has been loaded successfully.


import pandas as pd
import mysql.connector

conn = mysql.connector.connect(
    user = 'root',
    password = 'root',
    host = '127.0.0.1',
    port = 3306,
    database = 'the_kiran_academy_db'
)

print("connection successfull")
```

```python
st1 = 'select * from categories'
st2 = 'select * from products'
st3 = 'select * from sales'

categories = pd.read_sql(st1,conn)
products = pd.read_sql(st2,conn)
sales = pd.read_sql(st3,conn)
```

```python
# Task 2: Load Data from MySQL into Pandas
#     Connect Python to MySQL.
#     Read all three tables into Pandas DataFrames.
#     Display the first 5 records of each DataFrame.
#     Check the shape of each DataFrame.
print(categories.head())
print()
print(products.head())
print()
print(sales.head())

print()

print(categories.shape)
print(products.shape)
print(sales.shape)
```

```python
# Task 3: Merge DataFrames
#     Merge products and categories.
#     Merge the result with sales.
#     Create a common DataFrame named final_df.
#     Display the first 10 records of final_df.

df1 = pd.merge(products,categories,how='inner', on='cat_id')
print(df1)
print()

df2 = pd.merge(df1,sales,how='inner',on='product_id')
print(df2)
print()

final_df = df2[['product_id','product_name','price','category_name','sale_id','qty','sale_date']]
```

```python
final_df
```

```python
# Task 4: Create New Column
#     Create a column named total_sales.
#     Formula:
#     total_sales = price * qty

# Display the following columns:
#     product_name
#     category_name
#     price
#     qty
#     total_sales

final_df['total_sales'] = final_df['price'] + final_df['qty']
final_df[['product_name','category_name','price','qty','total_sales']]
```

```python
# Task 5: Aggregate Functions
# Perform the following calculations:
#     Total Sales
#     Average Sales
#     Maximum Sales
#     Minimum Sales
#     Total Quantity Sold
#     Average Product Price

print(final_df['total_sales'].sum())
print(final_df['total_sales'].mean())
print(final_df['total_sales'].max())
print(final_df['total_sales'].min())
print(final_df['qty'].sum())
print(final_df['price'].mean())
```

```python
# Task 6: Filtering
#     Display products with price greater than 10000.
#     Display records where quantity sold is greater than 5.
#     Display products from the Electronics category.
#     Display records where total_sales is greater than 50000.

print(final_df[final_df['price']>10000])
```

```python
print(final_df[final_df['qty']>5])
```

```python
# Task 7: GroupBy Analysis
#     Calculate total sales category-wise.

gp = final_df.groupby(by="category_name")

print(gp['total_sales'].sum())
print()

#     Calculate total quantity sold category-wise.
print(gp['qty'].sum())
print()

#     Calculate average price category-wise.
print(gp['price'].mean())
print()

#     Calculate total sales product-wise.
product_gp = final_df.groupby(by="product_name")
product_gp['total_sales'].sum()
print()

#     Find the category with the highest sales.
print(gp['total_sales'].sum())
print()
```

```python

# Task 8: Business Questions
#     Which product generated the highest sales?
#     Which category generated the highest sales?
#     What is the total revenue generated by the company?
#     Which product sold the highest quantity?
#     What is the average sales value per transaction?

print(final_df.loc[final_df['total_sales'].idxmax()])
print()

print(max(gp['total_sales']))
print()
```
