```python
# Pandas Election Data Analysis Task
# File 1: candidates.csv
# candidate_id	candidate_name	party
# C101	Rahul Sharma	BJP
# C102	Priya Patil	INC
# C103	Amit Verma	BJP
# C104	Neha Singh	AAP
# C105	Karan Gupta	INC

# File 2: constituencies.csv
# constituency_id	constituency_name	state
# S01	Nagpur Central	Maharashtra
# S02	Pune City	Maharashtra
# S03	Indore East	Madhya Pradesh
# S04	Delhi North	Delhi
# S05	Jaipur South	Rajasthan

# File 3: election_results.csv
# result_id	candidate_id	constituency_id	votes
# 1	C101	S01	85000
# 2	C102	S02	78000
# 3	C103	S03	NaN
# 4	C104	S04	65000
# 5	C105	S05	72000
# 5	C105	S05	72000
```

```python
import pandas as pd
```

```python
# Tasks
# Task 1: Load Data
#     Load all CSV files into Pandas DataFrames.

candidate_df = pd.read_csv('candidates.csv')
candidate_df
```

```python
constituencies_df = pd.read_csv('constituencies.csv')
constituencies_df
```

```python
election_result_df = pd.read_csv('election_results.csv')
election_result_df
```

```python

# Task 2: Basic Inspection
#     Display first 5 rows
#     Check shape
#     Check data types
#     Check column names
print(candidate_df.head())
print()
print(candidate_df.shape)
print()
print(candidate_df.dtypes)
print()
print(candidate_df.columns)
```

```python
# Task 3: Null Value Handling
#     Check Null Values
#     Replace Null Votes
#     Replace missing votes with average votes.
election_result_df.isnull().sum()
avg = election_result_df['votes'].mean()
election_result_df.fillna(avg,inplace=True)
election_result_df
```

```python
# Task 4: Duplicate Handling
#     Check Duplicates
#     Remove Duplicates
print(election_result_df.duplicated().sum())
print(constituencies_df.duplicated().sum())
print(candidate_df.duplicated().sum())
election_result_df.drop_duplicates()
```

```python
# Task 5: Merge Data

#     Merge all three files into one DataFrame.

#     Final Columns
#     candidate_name	party	constituency_name	state	votes

df1 = pd.merge(election_result_df,candidate_df, how="inner", on='candidate_id')
df1
```

```python
df2 = pd.merge(df1, constituencies_df, how="inner", on = 'constituency_id')
df2
```

```python
result = df2[['candidate_name','party','constituency_name','state','votes']]
result
```

```python
# Task 6: Apply Function
#     Create a column:
#     vote_category
#     Condition:
#     Votes > 80000       High
#     Votes > 70000       Medium
#     Else                Low
def cal_vote_category(vote):
    if int(vote)>80000:
        return 'High'
    elif int(vote)>70000:
        return 'Medium'
    else:
        return 'Low'
result['vote_category'] = result['votes'].apply(cal_vote_category)
result
```

```python
# Task 7: Filtering
# Find:
#     All BJP candidates
#     All INC candidates
#     Candidates from Maharashtra
#     Candidates with votes greater than 75000
#     Constituencies in Delhi
print(result[result['party'] == 'BJP'])
print()
print(result[result['party'] == 'INC'])
```

```python
result[result['state'] == 'Maharashtra']
```

```python
result[result['votes']>75000]
```

```python
result[result['constituency_name'] == 'Delhi']
```

```python
# Task 8: Aggregate Functions
# Calculate:
#     Total Votes
#     Average Votes
#     Maximum Votes
#     Minimum Votes
#     Count of Candidates

print(result['votes'].sum())
print(result['votes'].mean())
print(result['votes'].max())
print(result['votes'].min())
print(result['votes'].count())
```

```python
# Task 9: Grouping
#     Party-wise Total Votes
#     State-wise Total Votes
#     Party-wise Average Votes
#     State-wise Maximum Votes

party_gp = result.groupby(by='party')
print(party_gp['votes'].sum())
print()

state_gp = result.groupby(by='state')
print(state_gp['votes'].sum())
print()

print(party_gp['votes'].mean())
print()

print(state_gp['votes'].max())
```

```python
# Task 10: Sorting
#     Sort by Votes Descending
#     Sort by Candidate Name Ascending
result.sort_values(by='votes', ascending=False)
```

```python
result.sort_values(by='candidate_name')
```

```python
# Task 11: Advanced Analysis
# Find Winning Candidate from Each State

id = result.groupby(by='state')['votes'].idxmax()
winner = result.loc[id]
winner

# Hint:
# groupby('state')['votes'].idxmax()


```

```python
# Find Party with Highest Total Votes
# Hint:
#     groupby('party')['votes'].sum()

party_votes = result.groupby(by='state')['votes'].sum()
print(party_votes.idxmax(), party_votes.max())
```

```python
# Top 3 Candidates by Votes
# Hint:
#     nlargest()

print(result.nlargest(3,'votes'))
```

```python
# Bottom 3 Candidates by Votes
# Hint:
#     nsmallest()

print(result.nsmallest(3,'votes'))
```

```python
# Task 12: Business Questions
#     Which party received the highest total votes?
print(result.groupby(by='party')['votes'].sum().max())
print()

#     Which state recorded the highest votes?
print(result.groupby(by='state')['votes'].sum().max())
print()

#     Who is the top-performing candidate?
print(result[result['votes']==result['votes'].max()])
print()

#     Which constituency had the highest votes?
print(result.loc[result.groupby(by='constituency_name')['votes'].idxmax()])
print()

#     What percentage of total votes did each party receive?
print(result.groupby(by='party'))
print()

#     Which party has the highest average votes per candidate?
avg_votes = result.groupby(by='party')['votes'].mean()
print(avg_votes.idxmax(), avg_votes.max())
print()

#     How many constituencies were won by each party?
print(result.loc[result.groupby('constituency_name')['votes'].idxmax()]['party'].value_counts())
```
