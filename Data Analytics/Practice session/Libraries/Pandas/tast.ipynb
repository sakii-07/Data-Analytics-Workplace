```python
# Task 1: Load CSV File

#     Create a CSV file named students.csv.

#     roll,name,course,fees,city
#     1,Amit,Python,25000,Pune
#     2,Neha,Data Science,45000,Mumbai
#     3,Rahul,Python,25000,Nagpur
#     4,Priya,Java,30000,Pune
#     5,Rohan,Data Analytics,40000,Nashik
#     6,Sneha,Java,30000,Mumbai
#     7,Karan,Python,25000,Pune
#     8,Pooja,Data Science,45000,Nagpur
#     9,Ajay,Data Analytics,40000,Pune
#     10,Meena,Java,30000,Nashik
# Load this file into a DataFrame.
```

```python
import pandas as pd
student_data = pd.read_csv("student_data.csv")
student_data
```

```python
# Task 2: Access Columns
#     Display only name column.
#     Display name and course columns.
#     Display fees column.
#     Find datatype of each column.

student_data['name']

student_data[['name','course']]

student_data['fees']

student_data.info()
```

```python
# Task 4: Add New Column
#     Add a column named GST.
#     GST = fees * 0.18
#     Expected:
#     fees	GST
#     25000	4500
#     45000	8100

student_data['GST'] = student_data['fees']*0.18

student_data
```

```python
# Task 5: Calculate New Column
#     Create:
#     total_fees = fees + GST
#     Output:
#     fees	GST	total_fees
#     25000	4500	29500

student_data['total_fees'] = student_data['fees'] + student_data['GST']
student_data
```

```python
# Task 6: Add New Row
# Add:
#     11,Vaibhav,AWS,35000,Pune
#     to the DataFrame.

student_data.loc[10] = ['11','Vaibhav','AWS','35000','Pune']
student_data
```

```python
# Task 7: Rename Columns
#     Rename:
#     name â†’ student_name
#     fees â†’ course_fee
#     city â†’ location

student_data.rename(columns={
    "name" : "student_name",
    'fees' : "course_fee",
    'city' : "location"
}, inplace=True)

student_data
```

```python
# Task 8: Filtering
#     Single Condition
#     Students from Pune.

student_data[student_data['location'] == 'Pune']

#     Students whose fees > 30000.

student_data[student_data['course_fee']>30000]

#     Multiple Conditions
#     Students from Pune AND fees > 25000.
student_data[(student_data['location'] == 'Pune') & (student_data['course_fee'] > 25000)]

#     Students from Mumbai OR course = Java.
student_data[(student_data['location'] == 'Mumbai') | (student_data['course'] == 'Java')]

#     Students from Nagpur AND course = Data Science.
student_data[(student_data['location'] == 'Nagpur') & (student_data['course'] == 'Data Science')]
```

```python
# Task 9: Sorting
#     Sort by fees ascending.
#     Sort by fees descending.
#     Sort by city and fees.

student_data.sort_values(by="course_fee")
student_data.sort_values(by="course_fee", ascending=False)
student_data.sort_values(by=['location','course_fee'])
```

```python
# Task 10: Create Another CSV

#     payments.csv

payment = pd.read_csv('payments.csv')
payment
```

```python
# Task 11: Merge DataFrames

# Load both files and perform:

#     ALL join operations

pd.merge(student_data,payment,how="inner", on="roll")

pd.merge(student_data,payment,how="left", on="roll")

pd.merge(student_data,payment,how="right", on="roll")

student = pd.merge(student_data,payment,how="outer", on="roll")
student
```

```python
# Task 12: Remaining Fees Calculation
# After merge create:
# remaining_fees = fees - paid_fees

# Display:
# name	fees	paid_fees	remaining_fees

student['remaining_fees'] = student['fees'] - student['paid_fees']
student[['name','fees','paid_fees','remaining_fees']]
```

```python
# Task 13: Complex Filtering
#     Students whose remaining fees > 5000.
#     Students from Pune and remaining fees > 5000.
#     Students enrolled in Python and paid less than 20000.
#     Students from Mumbai or Nagpur with fees > 30000.

student[student['remaining_fees']>5000]

student[(student['city'] == 'Pune') & (student['remaining_fees']>5000)]

student[(student['course'] == 'Python') & (student['paid_fees']<20000)]

student[(student['city'].isin(['Mumbai','Nagpur'])) & (student['paid_fees']<30000)]
```
