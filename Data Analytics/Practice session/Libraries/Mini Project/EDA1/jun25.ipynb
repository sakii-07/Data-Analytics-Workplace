```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# <!-- Task 1: Data Loading
#     Load all three CSV files.
#     Display the first and last five records.
#     Check the shape, columns, and data types of each dataset. -->
```

```python
student_df = pd.read_csv('students.csv')
student_df.head()
```

```python
course_df = pd.read_csv('courses.csv')
course_df.head()
```

```python
trainer_df = pd.read_csv('trainers.csv')
trainer_df.head()
```

```python
#     Check the shape, columns, and data types of each dataset. -->
print("shape : ",student_df.shape)
print()
print("columns : ",student_df.columns)
print()
print("data type : \n",student_df.dtypes)
```

```python
print("shape : ",course_df.shape)
print()
print("columns : ",course_df.columns)
print()
print("data types : \n",course_df.dtypes)
```

```python
print("shape : ",trainer_df.shape)
print()
print("columns : ",trainer_df.columns)
print()
print("data type : \n",trainer_df.dtypes)
```

```python
# Task 2: Data Cleaning
#     Check and handle missing values.
#     Remove duplicate records.
#     Convert date columns into the correct format.
#     Correct inconsistent values if found.
```

```python
print("Null values count : \n",student_df.isnull().sum())
print()
print("Duplicate values count : ",student_df.duplicated().sum())
print()
student_df['Admission_Date'] = pd.to_datetime(student_df['Admission_Date'])
student_df.head()
```

```python
print("Null values count : \n",course_df.isnull().sum())
print()
print("Duplicate values count : ",course_df.duplicated().sum())
print()
course_df['Start_Date'] = pd.to_datetime(course_df['Start_Date'])
course_df.head()
```

```python
print("Null values count : \n",trainer_df.isnull().sum())
print()
print("Duplicate values count : ",trainer_df.duplicated().sum())
print()
trainer_df['Joining_Date'] = pd.to_datetime(trainer_df['Joining_Date'])
trainer_df.head()
```

```python
# Task 3: Data Analysis
#     Merge all three datasets.

df1 = pd.merge(student_df, course_df, how="inner", on="Course_ID")

df2 = pd.merge(df1, trainer_df, how="inner", on="Trainer_ID")
df2.head()
```

```python
#     Find the total number of students, courses, and trainers.

total_student = df2['Student_ID'].count()
print("Total students : ",total_student)

total_courses = df2['Course_ID'].count()
print("Total courses : ",total_courses)

total_trainers = df2['Trainer_ID'].count()
print("Total trainers : ",total_trainers)
```

```python
#     Find the most popular course.

df2.groupby(by="Course_Name")['Course_Name'].value_counts().sort_values(ascending=False).head(1)
```

```python
#     Find the branch with the highest admissions.

df2.groupby(by="Branch_x")['Branch_x'].value_counts().sort_values(ascending=False).head(1)
```

```python
#     Find the average marks for each course.
df2.groupby(by="Course_Name")['Marks'].mean()
```

```python
#     Find the total revenue generated by each branch.
df2.groupby(by="Branch_x")['Fees_Paid'].sum()
```

```python
#     Find the trainer handling the maximum number of students.
df2.groupby(by="Course_Name")['Trainer_Name'].value_counts().sort_values(ascending=False).head(1)
```

```python
#  Find the placement percentage for each course.
status = df2.groupby(by="Course_Name")['Placement_Status'].value_counts()
status
```

```python
# Task 4: Data Visualization
#     Create suitable visualizations for:
#     Student count by course

sns.countplot(data=df2, x='Course_Name')
plt.show()
```

```python
#     Revenue by branch

sns.barplot(data=df2, x='Branch_x', y='Fees_Paid')
plt.show()
```

```python
#     Average marks by course

sns.barplot(data=df2, x="Course_Name",y='Marks')
```

```python
#     Placement status

sns.countplot(df1, x='Placement_Status', hue='Placement_Status')
plt.show()
```

```python
#     Attendance vs Marks (Scatter Plot)

sns.scatterplot(df2, x='Attendance', y='Marks',hue='Marks')
plt.show()
```

```python
#     Correlation Heatmap
corr = df2[['Attendance', 'Marks']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
# Task 5: Business Insights
# Based on your analysis, answer the following:
#     Which course is the most popular?
print(df2.groupby(by='Course_Name')['Course_Name'].value_counts().idxmax())
print()



#     Which branch generates the highest revenue?
print(df2.groupby(by='Branch_x')['Fees_Paid'].sum().max())
print()

#     Which trainer handles the most students?
print(df2.groupby(by='Trainer_Name')['Trainer_Name'].value_counts().idxmax())
print()

#     Does higher attendance lead to better marks?
correlation = df2['Attendance'].corr(df2['Marks'])
print("Correlation:", correlation)

#     Give 5 business recommendations based on your findings.
```
